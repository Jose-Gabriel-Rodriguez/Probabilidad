# Ejercicio 1

La probabilidad de éxito de una vacuna es de 0.83. Calcular la probabilidad de que una vez administrada a 16 pacientes **NINGUNO** sufra la enfermedad.

- N = **16** número de pacientes a los que se administra la vacuna  
- A = éxito de la vacuna
- p = P(A) = 0.83, por lo tanto, q = 1 - 0.83   q = 0.17
- X = **16** número de pacientes que no sufren la enfermedad



**Fuente:** https://www.youtube.com/watch?v=sGfGVwiRs3I

```python
from scipy.stats import binom

# Parámetros de la distribución binomial
probabilidad_exito = 0.83
num_pacientes = 16
X = 16  # Numero de exitos

# Calcular la probabilidad de que ningún paciente sufra la enfermedad
probabilidad_ninguno_enfermo = binom.pmf( X, num_pacientes, probabilidad_exito)

print("La probabilidad de que ningún paciente sufra la enfermedad es:", probabilidad_ninguno_enfermo)
```

## Ejercicio 2

La probabilidad de éxito de una vacuna es de 0.83. Calcular la probabilidad de que una vez administrada a 16 pacientes **TODOS** sufran la enfermedad.

- N = **16** número de pacientes a los que se administra la vacuna  
- A = éxito de la vacuna
- p = P(A) = 0.83, por lo tanto, q = 1 - 0.83   q = 0.17
- X = **0** número de pacientes que no sufren la enfermedad


**Fuente:** https://www.youtube.com/watch?v=sGfGVwiRs3I

```python
from scipy.stats import binom

# Parámetros de la distribución binomial
probabilidad_exito = 0.83
num_pacientes = 16
X = 0

# Calcular la probabilidad de que todos los pacientes sufran la enfermedad
probabilidad_todos_enfermos = binom.pmf(X, num_pacientes, probabilidad_exito)

print("La probabilidad de que todos los pacientes sufran la enfermedad es:", probabilidad_todos_enfermos)

```

#### Conclusiones:
En el caso anterior la probabilidad es casi nula, debido principalmente a que la probabilidad de éxito de la vacuna es muy alta

## Ejercicio 3

- Para una variable aleatoria binomial con n =10 y una probabilidad de 0.10
- Encontrar la probabilidad de la variable aleatoria x = 2. Realizar un programa en python que resuelva el problema usando una distribucion de probabilidad binomial

Fuente: *libro de Introduccion a la probabilidad, pag. 187*

```python
from scipy.stats import binom

# Parámetros de la distribución binomial
n = 10
p = 0.10
x = 2

# Calcular la probabilidad de la variable aleatoria X = 2
probabilidad_x_2 = binom.pmf(x, n, p)

print("La probabilidad de que la variable aleatoria X sea igual a 2 es:", probabilidad_x_2)
```

## Ejercicio 3, parte 2

- Para una variable aleatoria binomial con n =10 y una probabilidad de 0.10.
- Calcular las probabilidades para cada valor de x = 0 hasta 10.
- Imprimir la tabla de valores de x y sus probabilidades.
- Graficar el histograma de la distribución binomial

Fuente: *libro de Introduccion a la probabilidad, pag. 187*

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros de la distribución binomial
n = 10
p = 0.10

# Calcular las probabilidades para cada valor de x de 0 a 10
valores_x = np.arange(0, n+1)
probabilidades_x = [binom.pmf(x, n, p) for x in valores_x]

# Imprimir la tabla de valores de x y sus probabilidades
print("Valores de x\tProbabilidades")
for x, probabilidad in zip(valores_x, probabilidades_x):
    print(f"{x}\t\t{probabilidad}")

# Graficar el histograma de la distribución binomial
plt.figure(figsize=(8, 6))
plt.bar(valores_x, probabilidades_x, color='skyblue')
plt.title('Distribución Binomial')
plt.xlabel('Valor de X')
plt.ylabel('Probabilidad')
plt.xticks(valores_x)
plt.grid(True)
plt.show()
```
