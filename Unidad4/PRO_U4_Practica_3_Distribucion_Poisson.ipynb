- **Aplica para variables aleatorias discretas, (número finito de valores).**
- Su distribución de probabilidad da un buen modelo para datos que **representa el número de sucesos de un evento específico en una unidad determinada de tiempo o espacio**.


- A continuación, algunos **ejemplos de experimentos** para los cuales la variable aleatoria x puede ser modelada por la variable aleatoria de Poisson:
1. El número de llamadas recibidas por un conmutador durante un tiempo determinado.
2. El número de bacterias por volumen pequeño de fluido.
3. El número de llegadas de clientes al mostrador de una caja de pago en un minuto determinado.
4. El número de descomposturas de una máquina durante un día determinado.
5. El número de accidentes de tránsito en un crucero dado durante un tiempo determinado.


- En cada uno de los ejemplos anteriores, **x representa el número de eventos que ocurren en un periodo**, durante el cual se puede esperar que ocurra un promedio de mu de estos eventos.
- Las únicas suposiciones necesarias, cuando se usa la distribución de Poisson para modelar experimentos tales como éstos, son que las cuentas o eventos ocurren al azar e independientemente unos de otros.



###  La fórmula para la distribución de probabilidad de Poisson, así como su media y varianza, se dan a continuación:



## Fórmula de probabilidad

$$
f(x) = \frac{{e^{ - \mu }\cdot \mu ^x }}{{x!}}
$$

en donde:

- $f(x)$ es la función de probabilidad para valores de $x=0,1,2,3..,n$
.

- $μ$ es el valor medio esperado en cierto lapso de tiempo. Algunas veces expresado como $λ$ lambda.

- $x$ es la variable aleatoria. Es una variable aleatoria discreta $(x=0,1,.2,...)$

- $e$ valor constante, es la base de los logaritmos naturales 2.71728

## Función scipy.stats.poisson.pmf

Para resolver este problema utilizando una distribución de probabilidad Poisson en Python, podemos usar la función **scipy.stats.poisson.pmf** para calcular la **función de masa de probabilidad (PMF)** de la **distribución de Poisson**.

# Ejercicio 1

El número promedio de accidentes de tránsito en cierto crucero de carretera es **dos por semana**. Suponga que el número de accidentes sigue una distribución de Poisson con una **media =  2**.

1. Encuentre la probabilidad de que no haya accidentes en este crucero de carretera durante un periodo de 1 semana.
2. Encuentre la probabilidad de que a lo sumo haya tres accidentes en esta sección de carretera durante un periodo de 2 semanas.


**Fuente:** Libro de introduccion a la probabilidad y estadistica pag. 200

```python
from scipy.stats import poisson

# Parámetros de la distribución de Poisson
media_accidentes = 2

# Parte 1: Probabilidad de que no haya accidentes en 1 semana
probabilidad_0_accidentes = poisson.pmf(0, media_accidentes)

print("1. Probabilidad de que no haya accidentes en 1 semana:", probabilidad_0_accidentes)

# Parte 2: Probabilidad de que a lo sumo haya tres accidentes en 2 semanas
# Convertimos la media a una media para 2 semanas
media_2_semanas = media_accidentes * 2

# Calculamos la probabilidad acumulada hasta 3 accidentes en 2 semanas
probabilidad_3_o_menos_accidentes = poisson.cdf(3, media_2_semanas)

print("2. Probabilidad de que a lo sumo haya tres accidentes en 2 semanas:", probabilidad_3_o_menos_accidentes)
```

# Ejercicio 2

- Una compañía de seguros de vida asegura las vidas de **5,000 hombres** de **42 años** de edad.
- Si estudios actuariales muestran que **la probabilidad de que cualquier hombre de 42 años muera en un año determinado es .001**, encuentre la probabilidad exacta de que la compañía **tendrá que pagar x = 4 reclamaciones durante un año determinado**.


**Fuente:** Libro de introduccion a la probabilidad y estadistica pag. 201

```python
from scipy.stats import poisson

# Parámetros de la distribución de Poisson
media_muertes = 5000 * 0.001  # 5000 hombres con una probabilidad de muerte de 0.001

# Número de reclamaciones
x = 4

# Calcular la probabilidad de que la compañía tenga que pagar 4 reclamaciones en un año determinado
probabilidad_4_reclamaciones = poisson.pmf(x, media_muertes)

print("La probabilidad de que la compañía tenga que pagar 4 reclamaciones es de:",
      probabilidad_4_reclamaciones)
```
